```python
from src.data_utils import load_data, train_test_split
from src.featurization import build_features
from src.model import train_rf

df = load_data("data/toy_assay.csv")
df_train, df_test = train_test_split(df)

X_train = build_features(df_train, rep="morgan", radius=2, n_bits=512)
X_test = build_features(df_test, rep="morgan", radius=2, n_bits=512)

y_train = df_train["activity"].values
y_test = df_test["activity"].values

clf = train_rf(X_train, y_train, seed=42)

y_prob = clf.predict_proba(X_test)[:, 1]
y_pred = (y_prob >= 0.5).astype(int)

print("y_test:", y_test)
print("y_prob:", y_prob)
print("y_pred:", y_pred)
```

```python
from src.eval import classification_metrics

metrics = classification_metrics(y_test, y_prob, threshold=0.5)
metrics
```
